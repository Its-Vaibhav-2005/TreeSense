```python
from pathlib import Path
from ultralytics import YOLO
import torch
import random
import cv2
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
weights_path = Path("Models/base/yolo11n.pt").resolve()
assert weights_path.exists(), f"Weights not found at {weights_path}"
model = YOLO(str(weights_path))
```

```python
dataset_root = Path("Tree-Top-View-1").resolve()
data_yaml = dataset_root / "data.yaml"
assert data_yaml.exists(), f"Data config not found at {data_yaml}"
```

```python
def seed_everything(seed: int = 42) -> None:
    """Seed python, numpy, and torch for reproducible experiments."""
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)


seed_everything(1234)
print("Reproducibility seed set to 1234")
```

### Training configuration with heavy augmentations

```python
augmentation_settings = {
    "degrees": 10.0,
    "translate": 0.1,
    "scale": 0.8,
    "shear": 8.0,
    "perspective": 0.0005,
    "flipud": 0.15,
    "fliplr": 0.5,
    "mosaic": 1.0,
    "mixup": 0.15,
    "copy_paste": 0.1,
    "hsv_h": 0.015,
    "hsv_s": 0.7,
    "hsv_v": 0.4,
    "erasing": 0.4,
}


def build_train_settings():
    project_dir = Path("runs/tree_canopy")
    project_dir.mkdir(parents=True, exist_ok=True)
    base_settings = {
        "data": str(data_yaml),
        "epochs": 200,
        "imgsz": 896,
        "batch": 16,
        "patience": 40,
        "optimizer": "AdamW",
        "lr0": 5e-3,
        "lrf": 5e-4,
        "momentum": 0.95,
        "weight_decay": 5e-4,
        "warmup_epochs": 5.0,
        "warmup_momentum": 0.8,
        "cos_lr": True,
        "close_mosaic": 15,
        "project": str(project_dir),
        "name": "yolo11s-tree-aug", 
        "device": 0,
        "cache": True,
    }
    base_settings.update(augmentation_settings)
    return base_settings
```

```python
train_settings = build_train_settings()
print("Training with settings:\n", train_settings)
train_results = model.train(**train_settings)
```

```python
best_weights = Path(train_results.save_dir) / "weights" / "best.pt"
assert best_weights.exists(), f"Best weights not found at {best_weights}"
best_model = YOLO(str(best_weights))
metrics = best_model.val(data=str(data_yaml), split="val", imgsz=896)
metrics.box.map, metrics.box.map50
```

### Visualize augmented predictions on validation samples

```python
val_image_dir = dataset_root / "valid" / "images"
if not val_image_dir.exists() or not any(val_image_dir.iterdir()):
    val_image_dir = dataset_root / "test" / "images"

sample_images = sorted(val_image_dir.glob("*.jpg"))[:4]
if not sample_images:
    raise FileNotFoundError(f"No validation/test images found inside {val_image_dir}")

fig, axes = plt.subplots(1, len(sample_images), figsize=(5 * len(sample_images), 5))
axes = np.atleast_1d(axes)

for ax, img_path in zip(axes, sample_images):
    prediction = best_model.predict(source=str(img_path), imgsz=896, conf=0.25, verbose=False)[0]
    plotted = prediction.plot()
    ax.imshow(cv2.cvtColor(plotted, cv2.COLOR_BGR2RGB))
    ax.set_title(img_path.name)
    ax.axis("off")

plt.tight_layout()
plt.show()
```

### Training history analytics

```python
def locate_results_csv(project_dir: Path = Path("runs/tree_canopy")) -> Path:
    """Return the most relevant Ultralytics results.csv for plotting."""
    if "train_results" in globals():
        candidate = Path(getattr(train_results, "save_dir", "")) / "results.csv"
        if candidate.exists():
            return candidate
    csv_paths = sorted(project_dir.glob("*/results.csv"), key=lambda p: p.stat().st_mtime, reverse=True)
    if not csv_paths:
        raise FileNotFoundError(
            f"No results.csv files found inside {project_dir}. Run training first to log metrics."
        )
    return csv_paths[0]


results_csv = locate_results_csv()
print(f"Using training log: {results_csv}")
training_history = pd.read_csv(results_csv)
training_history["epoch"] = training_history.index
training_history.head()
```

```python
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(3, 2, figsize=(18, 18))
axes = axes.ravel()

# 1. Model performance
axes[0].plot(training_history["epoch"], training_history["metrics/mAP50-95(B)"], label="mAP@50:95", linewidth=2)
axes[0].plot(training_history["epoch"], training_history["metrics/mAP50(B)"], label="mAP@50", linewidth=2)
axes[0].set_title("Model performance vs epochs")
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("mAP")
axes[0].legend()

# 2. Box loss
axes[1].plot(training_history["epoch"], training_history["train/box_loss"], color="#ff7f0e", linewidth=2)
axes[1].set_title("Box loss vs epochs")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Loss")

# 3. Class loss
axes[2].plot(training_history["epoch"], training_history["train/cls_loss"], color="#2ca02c", linewidth=2)
axes[2].set_title("Class loss vs epochs")
axes[2].set_xlabel("Epoch")
axes[2].set_ylabel("Loss")

# 4. Objectness loss
axes[3].plot(training_history["epoch"], training_history["train/obj_loss"], color="#d62728", linewidth=2)
axes[3].set_title("Objectness loss vs epochs")
axes[3].set_xlabel("Epoch")
axes[3].set_ylabel("Loss")

# 5. Conf placeholder replaced later, reuse for accuracy chart
axes[4].plot(training_history["epoch"], training_history["metrics/precision(B)"], color="#1f77b4", linewidth=2)
axes[4].set_title("Accuracy (Precision proxy) vs epochs")
axes[4].set_xlabel("Epoch")
axes[4].set_ylabel("Precision")

# 6. mAP@50:95 rolling accuracy? maybe metrics/recall? use recall for obj? but requirement accuracy. We'll plot recall to show detection accuracy complement.
axes[5].plot(training_history["epoch"], training_history["metrics/recall(B)"], color="#9467bd", linewidth=2)
axes[5].set_title("Recall vs epochs")
axes[5].set_xlabel("Epoch")
axes[5].set_ylabel("Recall")

for ax in axes:
    ax.grid(True, alpha=0.3)
    ax.set_xlim(training_history["epoch"].min(), training_history["epoch"].max())

plt.tight_layout()
plt.show()
```

### Confusion matrix from best weights

```python
if "metrics" not in globals() or metrics is None:
    best_dir = Path(train_results.save_dir) if "train_results" in globals() else Path(results_csv).parent
    best_weights = best_dir / "weights" / "best.pt"
    assert best_weights.exists(), f"Best weights not found at {best_weights}"
    best_model = YOLO(str(best_weights))
    metrics = best_model.val(data=str(data_yaml), split="val", imgsz=train_settings.get("imgsz", 896), conf=0.25, plots=False)

cm = metrics.confusion_matrix.matrix
class_names = list(metrics.names.values()) if isinstance(metrics.names, dict) else metrics.names

plt.figure(figsize=(8, 7))
sns.heatmap(cm, annot=True, fmt=".1f", cmap="Blues", xticklabels=class_names, yticklabels=class_names, cbar=True)
plt.title("Confusion matrix (validation)")
plt.xlabel("Predicted class")
plt.ylabel("True class")
plt.tight_layout()
plt.show()
```
